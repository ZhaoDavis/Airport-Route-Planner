```python
import numpy as np
import pandas as pd
```

```python
#https://openflights.org/data.php

input = pd.read_csv("airports.dat")
EU_df = input[input.iloc[:,11].str.startswith("Europe/", na=False)]
EU_df = EU_df.iloc[:, [1, 2, 3, 5, 6, 7]]

EU_df = EU_df.set_axis(["Airport Name", "City", "Country", "ICAO Code", "Latitude", "Longitude"], axis=1)

EU_df["Airport Name"] = EU_df["Airport Name"].replace(",", "", regex=True)
EU_df["City"] = EU_df["City"].replace(",", "", regex=True)

EU_df = EU_df.dropna()

EU_df[EU_df["Airport Name"] == "Harstad/Narvik Airport Evenes"]
```

```python
EU_df.to_csv("EU_airports.csv", index=False)
```
